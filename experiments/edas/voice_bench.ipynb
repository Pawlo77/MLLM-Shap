# EDA

```python
sample_ds = load_dataset("hlt-lab/voicebench", CONFIGS["AdvBench"])
```

```python
pprint(sample_ds)
```

```python
pprint(sample_ds["test"].info)
```

```python
# Get number of rows
num_rows = len(sample_ds["test"])

# Sum up audio durations in seconds and convert to minutes
total_seconds = sum(sample["audio"]["duration"] for sample in sample_ds["test"])
total_audio_minutes = total_seconds / 60

# Create and print stats dictionary
stats = {"num_rows": num_rows, "total_audio_minutes": total_audio_minutes}
print(stats)
```
