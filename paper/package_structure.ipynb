```python
import os
```

```python
def print_package_structure(root_dir: str, indent: str = "") -> None:
    """
    Recursively print a Python-style package structure starting at root_dir.
    Shows only directories and .py files.

    Args:
        root_dir: The root directory to start from.
        indent: Indentation for the current level (used in recursion).
    """
    entries = sorted(os.listdir(root_dir))
    for entry in entries:
        path = os.path.join(root_dir, entry)
        # Show directories (potential subpackages)
        if os.path.isdir(path) and not entry.startswith("__pycache__"):
            print(f"{indent}{entry}/")
            print_package_structure(path, indent + "    ")
        elif entry.endswith(".py") or entry.endswith(".rst"):
            print(f"{indent}|── {entry}")
```

```python
root = os.path.abspath(os.path.join("..", "mllm_shap"))
print_package_structure(root)
```
